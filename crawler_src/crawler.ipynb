```python
def get_valid_entries(har):
    valid = []
    for entry in har['log']['entries']:
        if entry['response']['status'] == 0:
            continue
        if entry['response'].get('redirectURL', '').startswith('chrome-extension://'):
            continue
        valid.append(entry)
    return valid
```
